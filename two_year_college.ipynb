```python
link = "https://raw.githubusercontent.com/humdskyline/data/main/CollegeScores2year.csv"
college_scores = Table.read_table(link)
college_scores.show(5)
```

college_scores.labels

### Some of the cells have null values which you see as nan


```python
all_locales = college_scores.group("Locale")
all_locales
```

### We need to get rid of those nulls before we group to get the unique values
******* If the column has Strings, nan will be the String "nan"

```python
non_null_locales = college_scores.where("Locale", are.not_equal_to("nan"))
locales = non_null_locales.group("Locale")
locales
```

```python
locales.barh("Locale")
```

### Since this is real data, there will be null values for some values for each state.  That is because some schools do not collect or report on some metrics.  Keep in mind that you will need to delete them before performing any function on them

### Let's look at the average completion rate for each state.
******* If the column has numbers, to get the rows that are NOT nan, use are.above(0)

### Write comments on how you would implement it first, and then write your code to do so

```python

non_null_comp_rates = college_scores.where("CompRate", are.above(0))
non_null_comp_rates

avg_by_state = non_null_comp_rates.group("State", np.average)
avg_comp_rate_by_state = avg_by_state.select("State", "CompRate average")
avg_comp_rate_by_state
```

```python
avg_comp_rate_by_state.barh("State")
```
