## Part 4: Explore Your Data (EDA)

Create 1–2 visualizations for each project variable that help you understand your variables or answer your research question.

### Option A: Histogram

Use this for **one numerical variable**. A histogram shows the distribution of the variable.

```python
# Histogram for one numerical variable
# Replace column_name with your numerical variable

plt.figure()

plt.hist(df["column_name"])

plt.xlabel("column_name")
plt.ylabel("Count")
plt.title("Histogram of column_name")

plt.show()
```

### Option B: Scatterplot

Use this for **two numerical variables**. A scatterplot helps you look for a relationship.

```python
# Scatterplot for two numerical variables
# Replace x_column and y_column with your variables

plt.figure()

plt.scatter(df["x_column"], df["y_column"])

plt.xlabel("x_column")
plt.ylabel("y_column")
plt.title("Scatterplot of y_column vs x_column")

plt.show()
```

### Option C: Bar Chart

Use this for **one categorical variable**. A bar chart shows how often each category appears.

```python
# Bar chart for one categorical variable
# Replace column_name with your categorical variable

plt.figure()

counts = df["column_name"].value_counts()
plt.bar(counts.index, counts.values)

plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Bar Chart of column_name")
plt.xticks(rotation=45)

plt.show()
```

### Option D: Compare Groups

Use this for **one numerical variable and one categorical variable**. This graph compares the average numerical value across groups.

```python
# Bar chart of group averages
# Replace category_column and numerical_column with your variables

plt.figure()

group_means = df.groupby("category_column")["numerical_column"].mean()
plt.bar(group_means.index, group_means.values)

plt.xlabel("Category")
plt.ylabel("Average value")
plt.title("Average numerical_column by category_column")
plt.xticks(rotation=45)

plt.show()
```
