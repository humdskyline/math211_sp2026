```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
link = "https://raw.githubusercontent.com/humdskyline/data/main/CollegeScores2year.csv"
scores_df = pd.read_csv(link)
scores_df.head()
```

## Categorial Data Graph: Bar Chart
Bar Chart: Average In-State Tuition by State

Recall groupby from last time

```python
# Group by 'State' and calculate the mean In-state Tuition
avg_tuition_by_state_gb = scores_df.groupby('State')['TuitionIn']
avg_tuition_by_state = avg_tuition_by_state_gb.mean()

# Create a bar plot to visualize the data
avg_tuition_by_state.plot(kind='bar', figsize=(12, 6), color='skyblue')
plt.title('Average In-State Tuition by State')
plt.xlabel('State')
plt.ylabel('Average Tuition ($)')
plt.xticks(rotation=90)  # Rotate x-axis labels for readability
plt.show()
```

Maybe it would be more useful if we ordered it by the in-state tuition

```python
# Now, let's sort the data by state from highest to lowest

avg_tuition_by_state_sorted = avg_tuition_by_state_gb.mean().sort_values(ascending=False)

# Create a bar plot to visualize the data
avg_tuition_by_state_sorted.plot(kind='bar', figsize=(12, 6), color='skyblue')
plt.title('Average In-State Tuition by State')
plt.xlabel('State')
plt.ylabel('Average Tuition ($)')
plt.xticks(rotation=90)  # Rotate x-axis labels for readability
plt.show()
```

## Numerical Data Graph: Histogram
Histogram of Distribution of Graduation Rates

```python
# National Completion (Graduation) Rate Column
comp_rates = scores_df["CompRate"]

#Get the average completion rate to overlay it on the histogram
avg_comp_rate = comp_rates.mean()
print(avg_comp_rate)


# Add a vertical line at the average completion rate
plt.axvline(avg_comp_rate, color='red', linestyle='dashed', linewidth=2)

# Plot the histogram of 'Completion Rate'
comp_rates.plot(kind='hist', bins=20, figsize=(10, 6), color='skyblue')
plt.title('Distribution of Completion Rates')
plt.xlabel('Completion Rate (%)')
plt.ylabel('Frequency')
plt.show()
```

Let's filter for California.  Recognize the filtering code from last time?

```python
# Filter for California Colleges
ca_colleges = scores_df[scores_df["State"]=="CA"]

# Completion Rates in California
ca_comp_rates = ca_colleges["CompRate"]

# Add a vertical line at the average completion graduation rate
plt.axvline(avg_comp_rate, color='red', linestyle='dashed', linewidth=2)

# Plot the histogram of 'Completion Rate'
ca_comp_rates.plot(kind='hist', bins=20, figsize=(10, 6), color='skyblue')
plt.title('Distribution of California Completion Rates')
plt.xlabel('Completion Rate (%)')
plt.ylabel('Frequency')
plt.show()
```

## Multiple numerical variable graph -- heatmap

Let's make a heatmap to see how different numerical variables are correlated

```python
# Select numeric columns for the heatmap
numeric_columns = ['TuitionIn', 'CompRate', 'PartTime', 'FullTimeFac']
df_numeric = scores_df[numeric_columns]

# Compute the correlation matrix
correlation_matrix = df_numeric.corr()

# Create a heatmap to visualize the correlation
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap of College Metrics')
plt.show()
```

It looks like in-state tuition has the highest correlation with completion rate.  Let's make a scatterplot of those two variables to take a closer look

```python
# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=scores_df, x="CompRate", y="TuitionIn", alpha=0.7, edgecolor="black")

# Customize the plot
plt.xlabel("Completion Rate")
plt.ylabel("In-State Tuition ($)")
plt.title("Scatter Plot of Completion Rate vs. In-State Tuition")
plt.grid(True)

# Show the plot
plt.show()
```
